# Credit Risk Prediction System

## Part A: Data acquisition and understanding

```python
import pandas as pd
import numpy as np

df = pd.read_csv("credit_risk_dataset.csv")

#Displaying shape
print("Dataset shape [rows, columns]:", df.shape)
                 
```

```python
#Displaying columns
print("Columns:")
print(df.columns.tolist())
```

```python
#Info
df.info()
```

```python
#describe
df.describe()
```

```python
#Identifying missing values
missing = df.isnull().sum()
missing_percent = (missing / len(df)) * 100
missing_summary = pd.DataFrame({'missing_count': missing, 'missing_percent': missing_percent})
missing_summary = missing_summary[missing_summary['missing_count'] > 0].sort_values('missing_count', ascending=False)
print("Columns with missing values:")
missing_summary
```

The dataset contains a mix of numeric and categorial features. And two columns `[loan_int_rate and person_emp_length]` contains missing values.

## PART B: Exploratory Data Analysis (EDA)

```python
!pip install seaborn

#Visualisations
import matplotlib.pyplot as plt #for creating charts/graphs
import seaborn as sns #for higher-level plots

sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (6, 4)
```

```python
#Target variable distribution
plt.figure()
axes = sns.countplot(x='loan_status', hue='loan_status', data = df,
                     palette=['lime', 'skyblue'], legend=False)
plt.title('Distribution of Target Variable: Loan Status')
plt.xlabel('Loan Status (0 = Low Risk / Repaid, 1 = High Risk / Default)') #X-axis label
plt.ylabel('Count') #y-axis label

for p in axes.patches:
    axes.annotate(f'{int(p.get_height())}', (p.get_x()+p.get_width()/2, p.get_height()),
                  ha='center', va='bottom') #horizontally centering the text and positioning text above the bar top
plt.tight_layout()
#Saving the chart as a png 
plt.savefig('01_target_distribution.png', dpi=110)
plt.show()

#Intepretation:
#The target classes are imbalanced (78% low-risk vs 22% high-risk).
#A naive model could reach 78% accuracy by just predicting the majority
# class, so techniques like class weighing and metric like
#precision/recal/F1 should be considered when modelling.
```

```python
#Correlation heatmap
plt.figure(figsize=(9, 7))
numeric_df = df.select_dtypes(include=[np.number])
corr = numeric_df.corr()
sns.heatmap(corr, annot=True, fmt='.2f',cmap='Blues', center=0)
plt.title('Correlation Heatmap of Numeric Features')
plt.tight_layout()
plt.savefig('02_correlation_heatmap.png', dpi=110)
plt.show()

#Interpretation:
#loan_percent_income and loan_int_rate show the strongest correlation with
#loan_status, suggesting they are useful predictors of default risk.
#perdon_age and cb_person_cred_hist_length are highly correlated (0.86),
#showing that redundant iformation [one could be dropped to reduce multicollinearity].
#Most other features show weak correlation with each other.
```

```python
#Boxplots ofr outliers
numeric_cols =['person_age', 'person_income', 'person_emp_length', 'loan_amnt',
              'loan_int_rate', 'loan_percent_income', 'cb_person_cred_hist_length']
fig, axes = plt.subplots(2, 4, figsize=(18, 8))
axes = axes.flatten()
for i, col in enumerate(numeric_cols):
    sns.boxplot(y=df[col], ax=axes[i], color='lightblue')
    axes[i].set_title(col)
axes[-1].axis('off')
plt.tight_layout()
plt.savefig('03_boxplot_outliers.png', dpi=110)
plt.show()

#Interpretation
#Boxplots reveal significant outliers across several features, some of which
#appear to be data entry errors rather then genuine values (e.g person_age
#reaching 144 and person_emp_length reaching 123 years are not realistic).
#person_income also has extreme outliers (up to 6 000 000) far above the
#typical range. loan_amnt, loan_int_rate, and loan_percent_income show more 
#convincing but still notable outliers and likely reflecting higher-risk borrowers.
#These outliers should be investigated and addressed (e.g capped, removed,
#or treated as missing) before modelling to avoid skewing results.
```
